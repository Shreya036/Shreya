```python
plt.boxplot(df["Fare"],patch_artist=True)
```

```python
print(f"Old shape of data:[df.Shape]")
```

```python
Q1=df["Fare"].quantile(0.25)
Q3=df["Fare"].quantile(0.75)
IQR=Q3-Q1
print(Q1,Q3,IQR)
```

```python
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
upper_index=np.where(df["Fare"]>=upper)[0]
lower_index=np.where(df["Fare"]<=lower)[0]
df.drop(index=upper_index,inplace=True)
df.drop(index=lower_index,inplace=True)
print(f"New shape of data:{df.shape}")
```
