```python
a=df["sepal_length"][:10]
b=df["sepal_width"][:10]
plt.figure(figsize=(7,5))
plt.plot(a,b)
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
plt.title('sepal_length vs sepal_width')
```

```python
a=df["sepal_length"][:10]
b=df["sepal_width"][:10]
c=range(10)
plt.plot(c,a,color="brown",label="sepal_length")
plt.plot(c,b,color="blue",label="sepal_width")
plt.xlabel("no.")
plt.ylabel("sepal dimensions")
plt.title("multiline plot")
plt.legend()
```
